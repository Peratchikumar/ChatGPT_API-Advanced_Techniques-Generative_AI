```python
#1.ROLE PROMPTING - TELL THE LLM WHAT ROLE IT SHOULD ACT AS

import os
# import → Keyword used to bring a library into the notebook
# os → Built-in library for interacting with the operating system (e.g., reading environment variables)

from dotenv import load_dotenv
# from → Keyword to import from a specific library
# dotenv → Library (python-dotenv) for reading key-value settings from a .env file
# import → Keyword used to bring in the specified item
# load_dotenv → Function that loads variables from a .env file into the environment

from openai import OpenAI
# from → Keyword to import from a specific library
# openai → Library used to connect to OpenAI's models
# import → Keyword used to bring in the specified item
# OpenAI → Class used to create a client that sends requests to OpenAI's API

load_dotenv(dotenv_path=".env")
# load_dotenv → Function that loads variables from a .env file into the environment
# (dotenv_path=".env") → Argument specifying the path/name of the file to read

openai_api_key = os.getenv("OPENAI_API_KEY")
# openai_api_key → Variable name storing the result
# = → Assignment operator
# os → Built-in operating system library
# .getenv → Function that reads the value of an environment variable
# ("OPENAI_API_KEY") → Argument specifying the name of the variable to read (your secret API key)

client = OpenAI()
# client → Variable name storing the result
# = → Assignment operator
# OpenAI() → Creates the API client object (automatically uses the API key from the environment)

MODEL = "gpt-5.6-luna"
# MODEL → Variable name (capitals show it is a constant that should not change)
# = → Assignment operator
# "gpt-5.6-luna" → String naming the OpenAI model to use (cost-efficient, 1M+ token context, "nano tier" of the GPT-5.6 family)

question = input("Enter User Question: ")
# question → Variable name storing the result
# = → Assignment operator
# input → Built-in function that waits for the user to type something
# ("Enter User Question: ") → Argument specifying the message shown to the user as the prompt

prompt = f"""
You are an expert Python developer.

Answer the user's question as a professional Python developer.
Give a clear and simple explanation with an example when useful.

User Question:
{question}
"""
# prompt → Variable name storing the result
# = → Assignment operator
# f""" ... """ → Triple-quoted f-string: a multi-line string that can insert variable values inside { }
# You are an expert Python developer. → Role instruction telling the LLM which role to act as
# Answer the user's question as a professional Python developer. → Task instruction telling the LLM how to answer
# Give a clear and simple explanation with an example when useful. → Style instruction controlling the answer's format
# User Question: → Label marking where the user's question begins
# {question} → Placeholder replaced with the text the user typed

response = client.responses.create(
    model=MODEL,
    input=prompt
)
# response → Variable name storing the result
# = → Assignment operator
# client → The OpenAI client object created earlier
# .responses → Attribute giving access to the Responses API
# .create → Function that sends a request to the model and gets a reply
# ( → Opens the function's arguments
# model=MODEL → Argument specifying which model to use
# input=prompt → Argument passing the full prompt text to the model
# ) → Closes the function's arguments

print("\nResponse:")
# print → Built-in function that displays output
# ("\nResponse:") → Argument specifying the text to display
# \n → Special character that adds a new line (blank line) before the text

print(response.output_text)
# print → Built-in function that displays output
# response → The reply object returned by the model
# .output_text → Attribute holding the model's answer as plain text

# This cell uses role prompting: it loads the API key from a .env file, asks the user for a question, tells the LLM to act as an expert 
# Python developer, sends the prompt to the gpt-5.6-luna model, and prints the model's answer.
```

```python
#3.XML PROMPTING - ORGANIZE THE PROMPT USING XML-STYLE TAGS

from openai import OpenAI
# from → Keyword to import from a specific library
# openai → Library used to connect to OpenAI's models
# import → Keyword used to bring in the specified item
# OpenAI → Class used to create a client that sends requests to OpenAI's API

client = OpenAI()
# client → Variable name storing the result
# = → Assignment operator
# OpenAI() → Creates the API client object (automatically uses the API key from the environment)

MODEL = "gpt-5.6-luna"
# MODEL → Variable name (capitals show it is a constant that should not change)
# = → Assignment operator
# "gpt-5.6-luna" → String naming the OpenAI model to use (cost-efficient, 1M+ token context, "nano tier" of the GPT-5.6 family)

user_query = input("Enter User Question: ")
# user_query → Variable name storing the result
# = → Assignment operator
# input → Built-in function that waits for the user to type something
# ("Enter User Question: ") → Argument specifying the message shown to the user as the prompt

prompt = f"""
<prompt>
    <role>Technical Assistant</role>
    <task>Provide clear, accurate, and practical technical solutions.</task>
    <user_query>{user_query}</user_query>
    <guidelines>
        <item>Use correct technical terminology.</item>
        <item>Be concise.</item>
        <item>Provide examples or code when relevant.</item>
        <item>Do not assume missing information.</item>
    </guidelines>
</prompt>
"""
# prompt → Variable name storing the result
# = → Assignment operator
# f""" ... """ → Triple-quoted f-string: a multi-line string that can insert variable values inside { }
# <prompt> ... </prompt> → Outer tag wrapping the entire prompt
# <role> ... </role> → Tag holding the role the LLM should act as
# Technical Assistant → The role given to the LLM
# <task> ... </task> → Tag holding what the LLM must do
# Provide clear, accurate, and practical technical solutions. → The task instruction
# <user_query> ... </user_query> → Tag holding the user's question
# {user_query} → Placeholder replaced with the text the user typed
# <guidelines> ... </guidelines> → Tag grouping the rules the LLM must follow
# <item> ... </item> → Tag holding one single rule inside the guidelines
# Use correct technical terminology. → Rule keeping the vocabulary accurate
# Be concise. → Rule keeping the answer short
# Provide examples or code when relevant. → Rule asking for examples or code where useful
# Do not assume missing information. → Rule stopping the LLM from guessing unknown details

response = client.responses.create(
    model=MODEL,
    input=prompt
)
# response → Variable name storing the result
# = → Assignment operator
# client → The OpenAI client object created earlier
# .responses → Attribute giving access to the Responses API
# .create → Function that sends a request to the model and gets a reply
# ( → Opens the function's arguments
# model=MODEL → Argument specifying which model to use
# input=prompt → Argument passing the full XML-structured prompt to the model
# ) → Closes the function's arguments

print("\nResponse:")
# print → Built-in function that displays output
# ("\nResponse:") → Argument specifying the text to display
# \n → Special character that adds a new line (blank line) before the text

print(response.output_text)
# print → Built-in function that displays output
# response → The reply object returned by the model
# .output_text → Attribute holding the model's answer as plain text

# This cell uses XML prompting: it asks the user for a question, organizes the role, task, query, and guidelines inside XML-style tags, 
# sends the structured prompt to the gpt-5.6-luna model, and prints the answer.
```

```python
#4.JSON PROMPTING - STRUCTURE THE PROMPT USING JSON

from openai import OpenAI
# from → Keyword to import from a specific library
# openai → Library used to connect to OpenAI's models
# import → Keyword used to bring in the specified item
# OpenAI → Class used to create a client that sends requests to OpenAI's API

import json
# import → Keyword used to bring a library into the notebook
# json → Built-in library for working with JSON data (converting between JSON text and Python dictionaries)

client = OpenAI()
# client → Variable name storing the result
# = → Assignment operator
# OpenAI() → Creates the API client object (automatically uses the API key from the environment)

MODEL = "gpt-5.6-luna"
# MODEL → Variable name (capitals show it is a constant that should not change)
# = → Assignment operator
# "gpt-5.6-luna" → String naming the OpenAI model to use (cost-efficient, 1M+ token context, "nano tier" of the GPT-5.6 family)

question = input("Enter User Question: ")
# question → Variable name storing the result
# = → Assignment operator
# input → Built-in function that waits for the user to type something
# ("Enter User Question: ") → Argument specifying the message shown to the user as the prompt

prompt = f"""
You are a technical assistant for IPython and Jupyter Notebook.

Your task is to answer the user's question directly and accurately.

Respond ONLY with valid JSON.

STRICT RULES:
1. Return exactly TWO key-value pairs.
2. The first key must be "topic".
3. The second key must be "description".
4. "topic" must contain the main topic of the user's question.
5. "description" must directly answer the user's question.
6. The description must contain ONE sentence with a maximum of 20 words.
7. Do not provide examples, lists, code, markdown, or explanations.
8. Do not discuss unrelated topics.
9. Do not change the user's question.
10. Return ONLY the JSON object.

Required format:
{{
  "topic": "topic name",
  "description": "short direct answer"
}}

User Question:
{question}
"""
# prompt → Variable name storing the result
# = → Assignment operator
# f""" ... """ → Triple-quoted f-string: a multi-line string that can insert variable values inside { }
# You are a technical assistant for IPython and Jupyter Notebook. → Role instruction setting the LLM's domain
# Your task is to answer the user's question directly and accurately. → Task instruction
# Respond ONLY with valid JSON. → Output-format instruction forcing a JSON-only reply
# STRICT RULES: → Label introducing the numbered rules the LLM must follow
# 1-3. Exactly TWO key-value pairs, first "topic", second "description" → Rules fixing the JSON structure and key order
# 4-5. "topic" holds the main topic, "description" answers the question → Rules defining what each key contains
# 6. ONE sentence with a maximum of 20 words → Rule limiting the answer length
# 7-9. No examples, lists, code, markdown, explanations, unrelated topics, or changed questions → Rules keeping the answer focused and clean
# 10. Return ONLY the JSON object → Rule stopping any extra text outside the JSON
# Required format: → Label introducing the exact output template
# {{ ... }} → Double curly braces: in an f-string, {{ prints a literal { and }} prints a literal }, so the JSON template is not mistaken for a variable
# "topic": "topic name" → Template line showing where the topic goes
# "description": "short direct answer" → Template line showing where the answer goes
# User Question: → Label marking where the user's question begins
# {question} → Placeholder replaced with the text the user typed (single braces = real variable)

response = client.responses.create(
    model=MODEL,
    input=prompt
)
# response → Variable name storing the result
# = → Assignment operator
# client → The OpenAI client object created earlier
# .responses → Attribute giving access to the Responses API
# .create → Function that sends a request to the model and gets a reply
# ( → Opens the function's arguments
# model=MODEL → Argument specifying which model to use
# input=prompt → Argument passing the full JSON-instruction prompt to the model
# ) → Closes the function's arguments

print("\nResponse:")
# print → Built-in function that displays output
# ("\nResponse:") → Argument specifying the text to display
# \n → Special character that adds a new line (blank line) before the text

print(response.output_text)
# print → Built-in function that displays output
# response → The reply object returned by the model
# .output_text → Attribute holding the model's answer as plain text (here, a JSON string)

# This cell uses JSON prompting: it asks the user for a question, instructs the LLM to reply only with a strict two-key JSON object 
# (topic and a one-sentence description under 20 words), sends the prompt to the gpt-5.6-luna model, and prints the JSON answer.
```

```python
#6.ReAct PROMPTING - THE PROMPT ENCOURAGES THE MODEL TO REASON THROUGH A PROBLEM STEP-BY-STEP BEFORE GIVING THE FINAL ANSWER

from openai import OpenAI
# from → Keyword to import from a specific library
# openai → Library used to connect to OpenAI's models
# import → Keyword used to bring in the specified item
# OpenAI → Class used to create a client that sends requests to OpenAI's API

client = OpenAI()
# client → Variable name storing the result
# = → Assignment operator
# OpenAI() → Creates the API client object (automatically uses the API key from the environment)

MODEL = "gpt-5.6-luna"
# MODEL → Variable name (capitals show it is a constant that should not change)
# = → Assignment operator
# "gpt-5.6-luna" → String naming the OpenAI model to use (cost-efficient, 1M+ token context, "nano tier" of the GPT-5.6 family)

question = input("Enter User Question: ")
# question → Variable name storing the result
# = → Assignment operator
# input → Built-in function that waits for the user to type something
# ("Enter User Question: ") → Argument specifying the message shown to the user as the prompt

prompt = f"""
You are an Technical assistant.

Solve the user's problem using this process:

1. Understand the problem.
2. Identify the important information.
3. Determine the best approach.
4. Check the result.
5. Give the final answer.

Do not provide hidden chain-of-thought.
Instead, provide only a short reasoning summary followed by
the final answer.

User Question:
{question}

Format:

Reasoning Summary:
<short explanation>

Final Answer:
<answer>
"""
# prompt → Variable name storing the result
# = → Assignment operator
# f""" ... """ → Triple-quoted f-string: a multi-line string that can insert variable values inside { }
# You are an Technical assistant. → Role instruction telling the LLM who it is
# Solve the user's problem using this process: → Instruction introducing the step-by-step method
# 1. Understand the problem. → Step 1: read and grasp what is being asked
# 2. Identify the important information. → Step 2: pick out the key facts needed
# 3. Determine the best approach. → Step 3: choose the best way to solve it
# 4. Check the result. → Step 4: verify the answer is correct
# 5. Give the final answer. → Step 5: state the conclusion
# Do not provide hidden chain-of-thought. → Rule stopping the LLM from exposing its full internal thinking
# Instead, provide only a short reasoning summary followed by the final answer. → Rule setting the output to a brief summary plus the answer
# User Question: → Label marking where the user's question begins
# {question} → Placeholder replaced with the text the user typed
# Format: → Label introducing the exact output layout
# Reasoning Summary: <short explanation> → Template section where the brief reasoning goes
# Final Answer: <answer> → Template section where the final result goes

response = client.responses.create(
    model=MODEL,
    input=prompt
)
# response → Variable name storing the result
# = → Assignment operator
# client → The OpenAI client object created earlier
# .responses → Attribute giving access to the Responses API
# .create → Function that sends a request to the model and gets a reply
# ( → Opens the function's arguments
# model=MODEL → Argument specifying which model to use
# input=prompt → Argument passing the full reasoning prompt to the model
# ) → Closes the function's arguments

print("\nResponse:")
# print → Built-in function that displays output
# ("\nResponse:") → Argument specifying the text to display
# \n → Special character that adds a new line (blank line) before the text

print(response.output_text)
# print → Built-in function that displays output
# response → The reply object returned by the model
# .output_text → Attribute holding the model's answer as plain text

# This cell uses ReAct-style prompting: it asks the user for a question, guides the LLM through a 5-step reasoning process, 
# requests only a short reasoning summary plus the final answer, sends the prompt to the gpt-5.6-luna model, and prints the result.
```

```python
#8.SKELETON OF THOUGHTS - FIRST CREATE A ROUGH STRUCTURE/SKELETON, THEN EXPAND IT INTO THE FINAL ANSWER

from openai import OpenAI
# from → Keyword to import from a specific library
# openai → Library used to connect to OpenAI's models
# import → Keyword used to bring in the specified item
# OpenAI → Class used to create a client that sends requests to OpenAI's API

client = OpenAI()
# client → Variable name storing the result
# = → Assignment operator
# OpenAI() → Creates the API client object (automatically uses the API key from the environment)

MODEL = "gpt-5.6-luna"
# MODEL → Variable name (capitals show it is a constant that should not change)
# = → Assignment operator
# "gpt-5.6-luna" → String naming the OpenAI model to use (cost-efficient, 1M+ token context, "nano tier" of the GPT-5.6 family)

question = input("Enter User Question: ")
# question → Variable name storing the result
# = → Assignment operator
# input → Built-in function that waits for the user to type something
# ("Enter User Question: ") → Argument specifying the message shown to the user as the prompt

prompt = f"""
You are an expert assistant.

Answer the user's question using Skeleton-of-Thought prompting.

Step 1:
Create a short skeleton containing the main points.

Step 2:
Expand each point into a useful explanation.

Do not provide hidden chain-of-thought.
Only provide the concise skeleton and its expanded answer.

User Question:
{question}

Format:

Skeleton:
1. <main point>

2. <main point>

3. <main point>

Expanded Answer:
<complete answer>
"""
# prompt → Variable name storing the result
# = → Assignment operator
# f""" ... """ → Triple-quoted f-string: a multi-line string that can insert variable values inside { }
# You are an expert assistant. → Role instruction telling the LLM who it is
# Answer the user's question using Skeleton-of-Thought prompting. → Instruction naming the technique to apply
# Step 1: Create a short skeleton containing the main points. → Stage 1: build the outline of key points
# Step 2: Expand each point into a useful explanation. → Stage 2: turn every outline point into a full explanation
# Do not provide hidden chain-of-thought. → Rule stopping the LLM from exposing its full internal thinking
# Only provide the concise skeleton and its expanded answer. → Rule limiting the output to the outline plus the expanded answer
# User Question: → Label marking where the user's question begins
# {question} → Placeholder replaced with the text the user typed
# Format: → Label introducing the exact output layout
# Skeleton: 1. / 2. / 3. <main point> → Template section listing three outline points
# Expanded Answer: <complete answer> → Template section where the full explanation goes

response = client.responses.create(
    model=MODEL,
    input=prompt
)
# response → Variable name storing the result
# = → Assignment operator
# client → The OpenAI client object created earlier
# .responses → Attribute giving access to the Responses API
# .create → Function that sends a request to the model and gets a reply
# ( → Opens the function's arguments
# model=MODEL → Argument specifying which model to use
# input=prompt → Argument passing the full Skeleton-of-Thought prompt to the model
# ) → Closes the function's arguments

print("\nResponse:")
# print → Built-in function that displays output
# ("\nResponse:") → Argument specifying the text to display
# \n → Special character that adds a new line (blank line) before the text

print(response.output_text)
# print → Built-in function that displays output
# response → The reply object returned by the model
# .output_text → Attribute holding the model's answer as plain text

# This cell uses Skeleton-of-Thought prompting: it asks the user for a question, instructs the LLM to first outline three main points 
# and then expand them into a complete answer, sends the prompt to the gpt-5.6-luna model, and prints the structured result.
```

```python
#9.PROMPT CHAINING - USE THE OUTPUT OF ONE LLM PROMPT AS THE INPUT TO ANOTHER PROMPT

from openai import OpenAI
# from → Keyword to import from a specific library
# openai → Library used to connect to OpenAI's models
# import → Keyword used to bring in the specified item
# OpenAI → Class used to create a client that sends requests to OpenAI's API

client = OpenAI()
# client → Variable name storing the result
# = → Assignment operator
# OpenAI() → Creates the API client object (automatically uses the API key from the environment)

MODEL = "gpt-5.6-luna"
# MODEL → Variable name (capitals show it is a constant that should not change)
# = → Assignment operator
# "gpt-5.6-luna" → String naming the OpenAI model to use (cost-efficient, 1M+ token context, "nano tier" of the GPT-5.6 family)

question = input("Enter User Question: ")
# question → Variable name storing the result
# = → Assignment operator
# input → Built-in function that waits for the user to type something
# ("Enter User Question: ") → Argument specifying the message shown to the user as the prompt

# CHAIN 1: Generate Explanation → Comment marking the first step, which gathers the key concepts

prompt_1 = f"""
Explain the following user question.

Give the key concepts and important points.
Do not write the final polished answer yet.

User Question:
{question}
"""
# prompt_1 → Variable name storing the result
# = → Assignment operator
# f""" ... """ → Triple-quoted f-string: a multi-line string that can insert variable values inside { }
# Explain the following user question. → Task instruction for Chain 1
# Give the key concepts and important points. → Instruction asking for the raw material of the answer
# Do not write the final polished answer yet. → Rule keeping Chain 1 from producing the finished answer
# User Question: → Label marking where the user's question begins
# {question} → Placeholder replaced with the text the user typed

response_1 = client.responses.create(
    model=MODEL,
    input=prompt_1
)
# response_1 → Variable name storing the result of Chain 1
# = → Assignment operator
# client → The OpenAI client object created earlier
# .responses → Attribute giving access to the Responses API
# .create → Function that sends a request to the model and gets a reply
# ( → Opens the function's arguments
# model=MODEL → Argument specifying which model to use
# input=prompt_1 → Argument passing the Chain 1 prompt to the model
# ) → Closes the function's arguments

intermediate_result = response_1.output_text
# intermediate_result → Variable name storing the result
# = → Assignment operator
# response_1 → The reply object returned by Chain 1
# .output_text → Attribute holding the model's answer as plain text (this is the link between the two chains)

# CHAIN 2: Create Final Answer → Comment marking the second step, which turns the explanation into a short answer

prompt_2 = f"""
Create a clear and simple final answer using the
information generated by the first prompt.

Original User Question:
{question}

Information from Chain 1:
{intermediate_result}

-Give only the final answer.
-Keep it short and simple.
-Maximum 5 sentences.
"""
# prompt_2 → Variable name storing the result
# = → Assignment operator
# f""" ... """ → Triple-quoted f-string: a multi-line string that can insert variable values inside { }
# Create a clear and simple final answer using the information generated by the first prompt. → Task instruction for Chain 2
# Original User Question: → Label marking the user's original question
# {question} → Placeholder replaced with the text the user typed
# Information from Chain 1: → Label marking the output passed on from Chain 1
# {intermediate_result} → Placeholder replaced with Chain 1's output (this is where the chaining happens)
# -Give only the final answer. → Rule stopping any extra explanation
# -Keep it short and simple. → Rule keeping the wording easy
# -Maximum 5 sentences. → Rule limiting the answer length

response_2 = client.responses.create(
    model=MODEL,
    input=prompt_2
)
# response_2 → Variable name storing the result of Chain 2
# = → Assignment operator
# client → The OpenAI client object created earlier
# .responses → Attribute giving access to the Responses API
# .create → Function that sends a request to the model and gets a reply
# ( → Opens the function's arguments
# model=MODEL → Argument specifying which model to use
# input=prompt_2 → Argument passing the Chain 2 prompt to the model
# ) → Closes the function's arguments

final_answer = response_2.output_text
# final_answer → Variable name storing the result
# = → Assignment operator
# response_2 → The reply object returned by Chain 2
# .output_text → Attribute holding the model's answer as plain text

print("\nResponse:")
# print → Built-in function that displays output
# ("\nResponse:") → Argument specifying the text to display
# \n → Special character that adds a new line (blank line) before the text

print(final_answer)
# print → Built-in function that displays output
# (final_answer) → Argument specifying what to display (only the final result, not the intermediate one)

# This cell uses prompt chaining: it asks the user for a question, sends it to the LLM to generate key concepts (Chain 1), feeds that output into a 
# second prompt to write a short final answer of up to 5 sentences (Chain 2), and prints only the final answer. 
# Both chains now use the gpt-5.6-luna model.
```

```python
#10.DYNAMIC TEMPLATE - DYNAMICALLY CHANGE THE PROMPT BASED ON USER INPUT

from openai import OpenAI
# from → Keyword to import from a specific library
# openai → Library used to connect to OpenAI's models
# import → Keyword used to bring in the specified item
# OpenAI → Class used to create a client that sends requests to OpenAI's API

client = OpenAI()
# client → Variable name storing the result
# = → Assignment operator
# OpenAI() → Creates the API client object (automatically uses the API key from the environment)

MODEL = "gpt-5.6-luna"
# MODEL → Variable name (capitals show it is a constant that should not change)
# = → Assignment operator
# "gpt-5.6-luna" → String naming the OpenAI model to use (cost-efficient, 1M+ token context, "nano tier" of the GPT-5.6 family)

question = input("Enter User Question: ")
# question → Variable name storing the result
# = → Assignment operator
# input → Built-in function that waits for the user to type something
# ("Enter User Question: ") → Argument specifying the message shown to the user as the prompt

style = input(
    "Enter Response Style "
    "(summarize / translate / explain/ code): "
)
# style → Variable name storing the result (the style the user picks)
# = → Assignment operator
# input → Built-in function that waits for the user to type something
# ( → Opens the function's arguments
# "Enter Response Style " → First part of the message shown to the user
# "(summarize / translate / explain/ code): " → Second part of the message, listing the allowed styles (two strings side by side are joined into one)
# ) → Closes the function's arguments

templates = {
    "summarize": "Summarize the following text in 3 concise bullet points.",
    "translate": "Translate the following text into French.",
    "explain": "Explain the following concept clearly.",
    "code": "Write only a short Python code for the following task. Do not explain the code"
}
# templates → Variable name storing the result
# = → Assignment operator
# { } → Curly braces creating a dictionary of key-value pairs
# "summarize": "Summarize the following text in 3 concise bullet points." → Key "summarize" mapped to an instruction for a 3-bullet summary
# "translate": "Translate the following text into French." → Key "translate" mapped to an instruction for French translation
# "explain": "Explain the following concept clearly." → Key "explain" mapped to an instruction for a clear explanation
# "code": "Write only a short Python code for the following task. Do not explain the code" → Key "code" mapped to an instruction for code-only output

selected_template = templates.get(
    style.lower(),
    templates["explain"]
)
# selected_template → Variable name storing the result (the chosen instruction)
# = → Assignment operator
# templates → The dictionary defined above
# .get → Function that looks up a key and returns a default value if the key is not found
# ( → Opens the function's arguments
# style.lower() → Argument: the user's input converted to lowercase, so "Explain" and "EXPLAIN" both match "explain"
# templates["explain"] → Argument: the default value returned when the style is not found
# ) → Closes the function's arguments

prompt = f"""
You are a helpful AI assistant.

{selected_template}

User Question:
{question}
"""
# prompt → Variable name storing the result
# = → Assignment operator
# f""" ... """ → Triple-quoted f-string: a multi-line string that can insert variable values inside { }
# You are a helpful AI assistant. → Role instruction telling the LLM who it is
# {selected_template} → Placeholder replaced with the instruction chosen from the dictionary (this is the dynamic part)
# User Question: → Label marking where the user's question begins
# {question} → Placeholder replaced with the text the user typed

response = client.responses.create(
    model=MODEL,
    input=prompt
)
# response → Variable name storing the result
# = → Assignment operator
# client → The OpenAI client object created earlier
# .responses → Attribute giving access to the Responses API
# .create → Function that sends a request to the model and gets a reply
# ( → Opens the function's arguments
# model=MODEL → Argument specifying which model to use
# input=prompt → Argument passing the dynamically built prompt to the model
# ) → Closes the function's arguments

print("\nResponse:")
# print → Built-in function that displays output
# ("\nResponse:") → Argument specifying the text to display
# \n → Special character that adds a new line (blank line) before the text

print(response.output_text)
# print → Built-in function that displays output
# response → The reply object returned by the model
# .output_text → Attribute holding the model's answer as plain text

# This cell uses a dynamic template: it asks the user for a question and a response style, picks the matching instruction from a dictionary 
# (defaulting to "explain" if the style is not recognized), builds the prompt with it, sends it to the gpt-5.6-luna model, and prints the answer.
```
