```python
#1.BUILDING A CUSTOM CHATBOT WITH CONVERSATION MEMORY

import os
# import → Keyword used to bring a library into the notebook
# os → Built-in library for interacting with the operating system (e.g., reading environment variables)

from dotenv import load_dotenv
# from → Keyword to import from a specific library
# dotenv → Library (python-dotenv) for reading key-value settings from a .env file
# import → Keyword used to bring in the specified item
# load_dotenv → Function that loads variables from a .env file into the environment

from openai import OpenAI
# from → Keyword to import from a specific library
# openai → Library used to connect to OpenAI's models
# import → Keyword used to bring in the specified item
# OpenAI → Class used to create a client that sends requests to OpenAI's API

load_dotenv(dotenv_path=".env")
# load_dotenv → Function that loads variables from a .env file into the environment
# (dotenv_path=".env") → Argument specifying the path/name of the file to read

openai_api_key = os.getenv("OPENAI_API_KEY")
# openai_api_key → Variable name storing the result
# = → Assignment operator
# os → Built-in operating system library
# .getenv → Function that reads the value of an environment variable
# ("OPENAI_API_KEY") → Argument specifying the name of the variable to read (your secret API key)

client = OpenAI()
# client → Variable name storing the result
# = → Assignment operator
# OpenAI() → Creates the API client object (automatically uses the API key from the environment)

messages = []
# messages → Variable name storing the result
# = → Assignment operator
# [] → Empty list that will hold the full conversation history

system_msg = input("What type of chatbot would you like to create?\n")
# system_msg → Variable name storing the result
# = → Assignment operator
# input → Built-in function that waits for the user to type something
# ("What type of chatbot would you like to create?\n") → Argument specifying the message shown to the user
# \n → Special character that adds a new line after the prompt text

messages.append({"role": "system", "content": system_msg})
# messages → The list being modified
# .append → Function that adds an item to the end of a list
# ({"role": "system" → Key-value pair marking this message as the system instruction
# "content": system_msg}) → Key-value pair holding the user's typed chatbot description as the system's behavior rule

print("Your new assistant is ready!")
# print → Built-in function that displays output
# ("Your new assistant is ready!") → Argument specifying the text to display

while True:
# while → Keyword starting a loop
# True → Condition that is always true, so the loop repeats until a break statement is reached
# : → Marks the start of the loop body

    message = input()
    # message → Variable name storing the result
    # = → Assignment operator
    # input → Built-in function that waits for the user to type something
    # () → Empty parentheses: no prompt message shown before typing

    if message.strip().lower() == "quit":
    # if → Keyword starting a conditional check
    # message → The text the user typed
    # .strip() → Function that removes leading/trailing spaces
    # .lower() → Function that converts the text to lowercase
    # == → Equality comparison operator
    # "quit" → String being compared against
    # : → Marks the start of the if-block

        print("Goodbye!")
        # print → Built-in function that displays output
        # ("Goodbye!") → Argument specifying the text to display

        break
        # break → Keyword that exits the while loop immediately

    messages.append({"role": "user", "content": message})
    # messages → The list being modified
    # .append → Function that adds an item to the end of a list
    # ({"role": "user" → Key-value pair marking this message as coming from the user
    # "content": message}) → Key-value pair holding what the user typed

    response = client.chat.completions.create(
        model="gpt-5.6-luna",
        messages=messages
    )
    # response → Variable name storing the result
    # = → Assignment operator
    # client → The OpenAI client object created earlier
    # .chat.completions → Attribute giving access to the Chat Completions API
    # .create → Function that sends a request to the model and gets a reply
    # ( → Opens the function's arguments
    # model="gpt-5.6-luna" → Argument specifying which model to use
    # messages=messages → Argument passing the entire conversation history so far (gives the chatbot memory)
    # ) → Closes the function's arguments

    reply = response.choices[0].message.content
    # reply → Variable name storing the result
    # = → Assignment operator
    # response → The reply object returned by the model
    # .choices → Attribute holding a list of possible responses (usually just one)
    # [0] → Indexing operator selecting the first (and only) response
    # .message → Attribute holding the message object of that response
    # .content → Attribute holding the actual answer text

    messages.append({"role": "assistant", "content": reply})
    # messages → The list being modified
    # .append → Function that adds an item to the end of a list
    # ({"role": "assistant" → Key-value pair marking this message as coming from the AI
    # "content": reply}) → Key-value pair storing the AI's reply so it remembers what it said

    print("\n" + reply + "\n")
    # print → Built-in function that displays output
    # "\n" → New line printed before the reply
    # + → String concatenation operator, joining pieces of text together
    # reply → The assistant's answer text
    # + "\n" → New line printed after the reply

# This cell builds a memory-enabled chatbot: it asks what kind of assistant to create, sets that as the system role, 
# then loops — taking user input, sending the full message history to gpt-5.6-luna, printing the reply, 
# and storing it back into memory for context in the next turn. Typing "quit" ends the loop cleanly.
```
