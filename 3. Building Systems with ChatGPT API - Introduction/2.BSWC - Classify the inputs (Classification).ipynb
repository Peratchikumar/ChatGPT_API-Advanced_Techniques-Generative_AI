```python
#1.INITIALIZE OPENAI API FOR INPUT CLASSIFICATION

import os
# import → Python keyword used to import a module
# os → Python module used to interact with the operating system

from dotenv import load_dotenv
# from → Imports a specific item from a module
# dotenv → Python package used to manage environment variables
# import → Python keyword used to import an item
# load_dotenv → Function used to load variables from the .env file

from openai import OpenAI
# from → Imports a specific item from a package
# openai → OpenAI Python package
# import → Python keyword used to import an item
# OpenAI → Class used to create an OpenAI API client

load_dotenv(dotenv_path=".env")
# load_dotenv → Loads environment variables from a file
# dotenv_path → Parameter that specifies the environment file path
# = → Assignment operator
# ".env" → Name of the environment file
# Overall → Loads the variables stored in the .env file

openai_api_key = os.getenv("OPENAI_API_KEY")
# openai_api_key → Variable used to store the API key
# = → Assignment operator
# os → Operating-system module
# getenv → Gets an environment variable
# "OPENAI_API_KEY" → Name of the API-key environment variable
# Overall → Retrieves the OpenAI API key

client = OpenAI()
# client → Variable used to store the OpenAI client
# = → Assignment operator
# OpenAI → Class used to create the API client
# () → Calls the OpenAI class
# Overall → Initializes the OpenAI API client


# EXPLANATION:
# This cell imports the required libraries,
# loads the OpenAI API key from the .env file,
# and initializes the OpenAI API client.

# BSW → Building Systems with ChatGPT API
# Evaluate Inputs → Processing user inputs
# Classification → Assigning inputs to specific categories

# This cell prepares the OpenAI API environment for the input classification task.
```

```python
#2.CREATE A REUSABLE CHATGPT COMPLETION FUNCTION

def get_completion_from_messages(messages,
                                 model="gpt-5.6-luna",
                                 temperature=1,
                                 max_tokens=500):
# def → Python keyword used to define a function
# get_completion_from_messages → Name of the function
# messages → Parameter containing the conversation messages
# model → Parameter specifying the AI model
# "gpt-5.6-luna" → Default model name used by the function
# temperature → Parameter controlling the randomness of the response
# 1 → Temperature value used for response generation
# max_tokens → Parameter controlling the maximum response length
# 500 → Maximum completion-token value
# : → Starts the function body


    response = client.chat.completions.create(
        model=model,
        messages=messages,
        temperature=temperature,
        max_completion_tokens=max_tokens,
    )
# response → Variable used to store the API response
# = → Assignment operator
# client → OpenAI API client created earlier
# . → Accesses an attribute or method
# chat → Accesses chat functionality
# completions → Accesses the completion-generation functionality
# create → Sends a request to create an AI response
# model=model → Sends the selected model to the API
# messages=messages → Sends the conversation messages to the API
# temperature=temperature → Sends the selected temperature value
# max_completion_tokens=max_tokens → Sets the maximum completion-token limit
# Overall → Sends the messages and generation settings to the OpenAI API


    return response.choices[0].message.content
# return → Sends a value back from the function
# response → Contains the API response
# choices → Contains the generated response choices
# [0] → Selects the first response choice
# message → Accesses the generated message
# content → Extracts the actual text response
# Overall → Returns the generated AI response as plain text


# EXPLANATION:
# This function creates a reusable connection to the OpenAI API.
# It accepts conversation messages and model-generation settings.
# The API generates a response using the selected model.
# The function then extracts and returns the generated message.

# messages → Conversation input sent to the model
# model → AI model selected for generation
# temperature → Controls response randomness
# max_tokens → Controls the maximum completion length
# response → Stores the API result
# choices → Contains generated responses
# content → Contains the final generated text

#This cell creates a reusable function for generating ChatGPT responses from a list of messages.
```

```python
#3.CLASSIFY CUSTOMER QUERIES USING CHATGPT API

delimiter = "####"
# delimiter → Variable used to store the separator
# = → Assignment operator
# "####" → Four hash characters used as the separator


system_message = f"""
You will be provided with customer service queries. \
The customer service query will be delimited with \
{delimiter} characters.
Classify each query into a primary category \
and a secondary category.
Provide your output in json format with the \
keys: primary and secondary.

Primary categories: Billing, Technical Support, \
Account Management, or General Inquiry.

Billing secondary categories:
Unsubscribe or upgrade
Add a payment method
Explanation for charge
Dispute a charge

Technical Support secondary categories:
General troubleshooting
Device compatibility
Software updates

Account Management secondary categories:
Password reset
Update personal information
Close account
Account security

General Inquiry secondary categories:
Product information
Pricing
Feedback
Speak to a human
"""

# system_message → Variable containing the instructions for the AI
# f → Creates an f-string
# """ → Starts and ends a multi-line string
# You → Refers to the AI model
# will be provided → Indicates that information will be given to the model
# customer service queries → Questions or requests from customers
# delimiter → Separator used to identify the customer query
# {delimiter} → Inserts the delimiter value into the string
# classify → Assign the query to a category
# each → Every individual query
# query → Customer question or request
# primary category → Main classification
# secondary category → More specific classification
# JSON → Structured data format
# keys → Names used to identify JSON values
# Billing → Payment-related category
# Technical Support → Technical problem category
# Account Management → Account-related category
# General Inquiry → General question category
# secondary categories → More specific classifications under each primary category


user_message = f"""
I want you to delete my profile and all of my user data
"""

# user_message → Variable containing the customer's query
# f → Creates an f-string
# """ → Defines a multi-line string
# I want → Expresses the customer's request
# delete → Permanently remove
# my profile → Customer account/profile
# all → Every item
# user data → Information associated with the customer


messages = [
    {'role': 'system',
     'content': system_message},

    {'role': 'user',
     'content': f"{delimiter}{user_message}{delimiter}"},
]

# messages → List containing the conversation messages
# [ ] → Creates a list
# { } → Creates dictionaries
# role → Defines the type of message
# system → Identifies the system instruction
# content → Contains the message text
# user → Identifies the customer input
# {delimiter} → Adds the separator before and after the query
# Overall → Combines the system instruction and customer query


response = get_completion_from_messages(messages)

# response → Variable storing the AI-generated classification
# = → Assignment operator
# get_completion_from_messages → Previously defined API function
# messages → Sends the system instruction and user query
# Overall → Sends the classification request to the AI model


print(response)

# print → Displays the result
# response → Contains the AI-generated classification
# Overall → Displays the classification result


# EXPLANATION:
# This cell classifies a customer service query into
# a primary category and a secondary category.
# The system message defines the available categories.
# The user query is placed between delimiter characters.
# The messages list sends both instructions and the query
# to the OpenAI API through the previously defined function.
# The final response displays the classification.

# This cell uses the ChatGPT API to classify customer queries into structured categories.
```
