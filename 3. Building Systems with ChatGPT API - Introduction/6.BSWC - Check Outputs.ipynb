````python
#1.IMPORTING LIBRARIES

# BSWC - Building Systems with ChatGPT API

# Check outputs


import os
# import → Python keyword used to load a module
# os → Python module used to interact with the operating system
# Overall → Imports the os module

from dotenv import load_dotenv
# from → Specifies the module from which something is imported
# dotenv → Python package used for environment variables
# import → Imports a specific function
# load_dotenv → Function that loads variables from a .env file
# Overall → Imports the function used to load environment variables

from openai import OpenAI
# from → Specifies the package from which something is imported
# openai → OpenAI Python package
# import → Imports a specific class
# OpenAI → OpenAI client class
# Overall → Imports the OpenAI API client

load_dotenv(dotenv_path=".env")
# load_dotenv → Loads environment variables from a file
# dotenv_path → Parameter specifying the .env file location
# ".env" → Name of the environment-variable file
# Overall → Loads the API key and other variables from .env

openai_api_key = os.getenv("OPENAI_API_KEY")
# openai_api_key → Variable used to store the OpenAI API key
# = → Assignment operator
# os → Python operating-system module
# getenv() → Retrieves an environment variable
# "OPENAI_API_KEY" → Name of the environment variable containing the API key
# Overall → Retrieves the OpenAI API key from the environment

client = OpenAI()
# client → Variable representing the OpenAI API client
# = → Assignment operator
# OpenAI() → Creates an OpenAI client object
# Overall → Initializes the client used to communicate with the OpenAI API


# EXPLANATION:
# This is the Check outputs section of the BSWC notebook.
# The cell imports the required Python packages,
# loads the environment variables from the .env file,
# retrieves the OPENAI_API_KEY, and initializes
# the OpenAI API client.

# This setup allows the following notebook cells
# to communicate with the OpenAI API.


# KEY CONCEPTS:
# BSWC → Building Systems with ChatGPT API
# CHECK OUTPUTS → Section used to verify notebook/API setup
# DOTENV → Package used to load environment variables
# OPENAI_API_KEY → Environment variable containing the API key
# CLIENT → Object used to communicate with the OpenAI API
# OpenAI() → Creates the OpenAI API client

# The OpenAI environment is initialized successfully.
````

````python
#2.GET COMPLETION FROM MESSAGES

def get_completion_from_messages(
    messages,
    model="gpt-5.6-luna",
    temperature=1,
    max_tokens=500
):
    # def → Python keyword used to define a function
    # get_completion_from_messages → Name of the function
    # messages → List containing the messages sent to the model
    # model → Parameter specifying the AI model
    # "gpt-5.6-luna" → Default model name
    # temperature → Controls the randomness of the model's response
    # 1 → Default temperature value
    # max_tokens → Parameter controlling the maximum completion-token limit
    # 500 → Default maximum-token value
    # : → Begins the function body
    # Overall → Defines a reusable function for generating model responses

    response = client.chat.completions.create(
        model=model,
        messages=messages,
        temperature=temperature,
        max_completion_tokens=max_tokens,
    )
    # response → Variable storing the API response
    # = → Assignment operator
    # client → OpenAI API client created earlier
    # chat → Chat API resource
    # completions → Completion-generation resource
    # create() → Sends a request to the API
    # model → Specifies which model to use
    # messages → Provides the conversation messages
    # temperature → Controls response randomness
    # max_completion_tokens → Sets the maximum number of completion tokens
    # max_tokens → Value supplied to max_completion_tokens
    # Overall → Sends the messages to the OpenAI API

    return response.choices[0].message.content
    # return → Sends a value back from the function
    # response → Contains the API response
    # choices → Collection of possible generated responses
    # [0] → Selects the first response
    # message → Accesses the message object
    # content → Accesses the generated text
    # Overall → Returns the model's generated response as text


# EXPLANATION:
# This function provides a reusable way to send messages
# to the OpenAI model and receive the generated response.
# Instead of writing the API request repeatedly,
# the function can be called whenever a model response is needed.
#
# The messages parameter contains the conversation.
# The model parameter specifies the model.
# temperature controls the randomness of the response.
# max_tokens controls the completion-token limit.
# The function finally returns the generated message content.


# KEY CONCEPTS:
# FUNCTION → Reusable block of Python code
# MESSAGES → Conversation sent to the model
# MODEL → AI model used to generate the response
# TEMPERATURE → Controls response randomness
# MAX_COMPLETION_TOKENS → Limits generated completion tokens
# RESPONSE → Object containing the API result
# CHOICES → Collection containing generated response options
# CONTENT → Actual text generated by the model

# This cell creates a reusable function for generating responses from the OpenAI API.
````

````python
#3.CHECK OUTPUT FOR POTENTIALLY HARMFUL CONTENT

final_response_to_customer = f"""
The SmartX ProPhone has a 6.1-inch display, 128GB storage, \
12MP dual camera, and 5G. The FotoSnap DSLR Camera \
has a 24.2MP sensor, 1080p video, 3-inch LCD, and \
interchangeable lenses. We have a variety of TVs, including \
the CineView 4K TV with a 55-inch display, 4K resolution, \
HDR, and smart TV features. We also have the SoundMax \
Home Theater system with 5.1 channel, 1000W output, wireless \
subwoofer, and Bluetooth. Do you have any specific questions \
about these products or any other products we offer?
"""
# final_response_to_customer → Variable containing the customer's final response
# = → Assignment operator
# f"""...""" → Formatted multi-line string
# SmartX ProPhone → Example product mentioned in the response
# 6.1-inch display → Product display specification
# 128GB storage → Product storage specification
# 12MP dual camera → Product camera specification
# 5G → Mobile network technology
# FotoSnap DSLR Camera → Example camera product
# 24.2MP sensor → Camera sensor specification
# 1080p video → Video recording specification
# 3-inch LCD → Camera display specification
# interchangeable lenses → Camera feature
# CineView 4K TV → Example television product
# 55-inch display → Television display size
# 4K resolution → Television resolution
# HDR → High Dynamic Range display technology
# smart TV features → Television software/connectivity features
# SoundMax Home Theater → Example home-theater product
# 5.1 channel → Surround-sound configuration
# 1000W output → Audio power specification
# wireless subwoofer → Wireless bass speaker
# Bluetooth → Wireless communication technology
# Overall → Creates the customer-response text that will be checked by moderation


response = client.moderations.create(
    model="omni-moderation-latest",
    input=final_response_to_customer
)
# response → Variable storing the moderation API response
# = → Assignment operator
# client → OpenAI API client created earlier
# moderations → OpenAI moderation API resource
# create() → Sends a moderation request
# model → Specifies the moderation model
# "omni-moderation-latest" → Current moderation model
# input → Text submitted for moderation
# final_response_to_customer → Customer response being evaluated
# Overall → Sends the customer response to the Moderation API


moderation_output = response.results[0]
# moderation_output → Variable storing the first moderation result
# = → Assignment operator
# response → Contains the moderation API response
# results → Collection containing moderation results
# [0] → Selects the first result
# Overall → Extracts the moderation result for the submitted text


print(moderation_output)
# print() → Python function used to display output
# moderation_output → Contains the moderation evaluation
# Overall → Displays the moderation result


# EXPLANATION:
# This cell checks a generated customer response for
# potentially harmful content using the OpenAI Moderation API.
# The complete customer response is stored in
# final_response_to_customer.
# The response is then submitted to the moderation model.
# The first moderation result is extracted and displayed.

# The moderation result can contain information such as
# whether the content was flagged and the moderation
# category results and scores.


# KEY CONCEPTS:
# MODERATION → Evaluating content for potentially harmful material
# FINAL RESPONSE → Customer-facing text being evaluated
# MODERATION MODEL → Model used to evaluate the submitted content
# INPUT → Text submitted to the moderation API
# RESULTS → Collection containing moderation results
# FLAGGED → Indicates whether the content was flagged
# CATEGORIES → Moderation categories evaluated by the API
# CATEGORY SCORES → Scores associated with moderation categories

# This cell checks the final customer response for potentially harmful content using the Moderation API.
````

````python
#4.CHECK IF OUTPUT IS FACTUALLY BASED ON THE PROVIDED PRODUCT INFORMATION

system_message = f"""
You are an assistant that evaluates whether \
customer service agent responses sufficiently \
answer customer questions, and also validates that \
all the facts the assistant cites from the product \
information are correct.
The product information and user and customer \
service agent messages will be delimited by \
3 backticks, i.e. ```.
Respond with Y or N character, with no punctuation:
Y - if the output sufficiently answers the question \
AND the response correctly uses product information
N - otherwise

Output a single letter only.
"""
# system_message → Variable containing the instructions for the model
# = → Assignment operator
# f"""...""" → Formatted multi-line string
# assistant → AI system being instructed
# evaluates → Checks or judges the provided response
# customer service agent → Person or system responding to the customer
# responses → Answers provided to customers
# sufficiently → Adequately or completely enough
# answer → Address the customer's question
# customer questions → Questions asked by the customer
# validates → Checks whether something is correct
# facts → Information or statements
# cites → Refers to or uses information
# product information → Information about the available products
# 3 backticks → Characters used to delimit the provided information
# Y → Indicates the response sufficiently answers the question and correctly uses product information
# N → Indicates the response does not satisfy the requirements
# no punctuation → Requires only Y or N without punctuation
# single letter → Requires exactly one classification letter
# Overall → Defines the rules for evaluating the customer-service response


customer_message = f"""
tell me about the smartx pro phone and \
the fotosnap camera, the dslr one. \
Also tell me about your tvs"""
# customer_message → Variable containing the customer's question
# = → Assignment operator
# f"""...""" → Formatted multi-line string
# tell me → Requests information
# smartx pro phone → Product the customer asks about
# fotosnap camera → Product the customer asks about
# dslr → Digital Single-Lens Reflex camera type
# Also → Adds another request
# TVs → Televisions
# Overall → Stores the customer's product-related question


product_information = """
{
    "name": "SmartX ProPhone",
    "category": "Smartphones and Accessories",
    "brand": "SmartX",
    "model_number": "SX-PRO10",
    "warranty": "1 year",
    "rating": 4.6,
    "features": ["6.1-inch display", "128GB storage", "12MP dual camera", "5G"],
    "description": "A powerful smartphone with advanced camera features.",
    "price": 899.99
}
"""
# product_information → Variable containing product details
# = → Assignment operator
# name → Product name
# category → Product category
# brand → Product manufacturer or brand
# model_number → Product identification number
# warranty → Warranty period
# rating → Product rating
# features → List of product features
# description → Brief description of the product
# price → Product selling price
# Overall → Stores the information used to verify the customer-service response


q_a_pair = f"""
Customer message: ```{customer_message}```
Product information: ```{product_information}```
Agent response: ```{final_response_to_customer}```

Does the response use the retrieved information correctly?
Does the response sufficiently answer the question?

Output Y or N
"""
# q_a_pair → Variable containing the complete question-and-answer evaluation prompt
# = → Assignment operator
# f"""...""" → Formatted multi-line string
# Customer message → The customer's original question
# {customer_message} → Inserts the customer's question
# Product information → Information used to verify factual accuracy
# {product_information} → Inserts the available product information
# Agent response → Customer-service response being evaluated
# {final_response_to_customer} → Inserts the previously generated customer response
# ``` → Delimits each section of information
# correctly → Checks whether the response accurately uses the product information
# sufficiently → Checks whether the response adequately answers the customer's question
# Output Y or N → Requires a binary classification
# Overall → Combines the customer question, product information, and agent response for evaluation


messages = [
    {'role': 'system', 'content': system_message},
    {'role': 'user', 'content': q_a_pair}
]
# messages → List containing messages sent to the model
# = → Assignment operator
# [ → Starts the list
# role → Identifies the type of message
# system → Identifies the system-level instructions
# content → Contains the actual message
# system_message → Provides the evaluation instructions
# user → Identifies the user message
# q_a_pair → Provides the evaluation information
# ] → Ends the list
# Overall → Creates the message structure for the model


response = get_completion_from_messages(messages, max_tokens=100)
# response → Variable storing the model's evaluation
# = → Assignment operator
# get_completion_from_messages() → Previously defined function for generating model responses
# messages → System and user messages sent to the model
# max_tokens → Maximum completion-token limit
# 100 → Provides sufficient output space for the model
# Overall → Sends the evaluation request to the model


print(response)
# print() → Displays the model's response
# response → Contains the Y or N classification
# Overall → Displays the final evaluation result


# EXPLANATION:
# This cell checks whether the customer-service response
# is factually supported by the provided product information.
# It evaluates two things:
# 1. Whether the response correctly uses the product information.
# 2. Whether the response sufficiently answers the customer's question.
#
# The model must return only Y or N.
# Y means both requirements are satisfied.
# N means at least one requirement is not satisfied.


# KEY CONCEPTS:
# FACTUAL VALIDATION → Checking whether statements are supported by the provided information
# PRODUCT INFORMATION → Reference information used to verify the response
# CUSTOMER MESSAGE → Question asked by the customer
# AGENT RESPONSE → Answer generated for the customer
# q_a_pair → Combined question, product information, and agent response
# DELIMITER → Boundary used to separate different pieces of information
# Y → Response is factually correct and sufficiently answers the question
# N → Response does not satisfy one or both requirements
# max_tokens=100 → Provides sufficient completion-token space

# This cell evaluates whether the customer-service response correctly uses the provided product information and sufficiently answers the 
# customer's question.
````
