```python
#1.CUSTOM CHATGPT WITH OPENAI API AND GRADIO

import os
# import → Python keyword used to import a module
# os → Python module used to interact with the operating system

import gradio
# import → Python keyword used to import a module
# gradio → Python library used to create web-based interfaces

from dotenv import load_dotenv
# from → Imports a specific item from a module
# dotenv → Python package used to manage environment variables
# import → Python keyword used to import an item
# load_dotenv → Function used to load variables from a .env file

from openai import OpenAI
# from → Imports a specific item from a package
# openai → OpenAI Python package
# import → Python keyword used to import an item
# OpenAI → Class used to create an OpenAI API client

load_dotenv(dotenv_path=".env")
# load_dotenv → Loads environment variables
# dotenv_path → Specifies the path of the environment file
# ".env" → Name of the environment-variable file
# Overall → Loads the API key stored in the .env file

openai_api_key = os.getenv("OPENAI_API_KEY")
# openai_api_key → Variable used to store the API key
# os → Operating-system module
# getenv → Gets an environment-variable value
# "OPENAI_API_KEY" → Name of the API-key environment variable
# Overall → Retrieves the OpenAI API key

client = OpenAI()
# client → Variable representing the OpenAI API client
# OpenAI() → Creates an OpenAI client object
# Overall → Creates the client used to communicate with the API

messages = [{"role": "system",
             "content": "You are a financial experts that specializes in real estate investment and negotiation"}]
# messages → Variable storing the conversation history
# [] → Creates a list
# {} → Creates a dictionary
# role → Defines the role of the message
# "system" → Identifies a system instruction
# content → Contains the instruction given to the AI
# "You are..." → Defines the AI's role and specialization
# Overall → Sets the initial behavior of the chatbot

def CustomChatGPT(user_input):
# def → Defines a function
# CustomChatGPT → Name of the function
# user_input → Parameter that receives user input
# ( ) → Contains the function parameter
# : → Starts the function block
# Overall → Defines the chatbot function

    messages.append({"role": "user", "content": user_input})
# messages → Conversation-history list
# append → Adds an item to the list
# role → Defines the message role
# "user" → Identifies the message as user input
# content → Contains the user's message
# user_input → Text entered by the user
# Overall → Adds the user's message to the conversation history

    response = client.chat.completions.create(
        model="gpt-5.6-luna",
        messages=messages
    )
# response → Variable storing the API response
# client → OpenAI API client
# chat → Accesses chat functionality
# completions → Accesses text-completion functionality
# create → Creates an API request
# model → Specifies the AI model
# "gpt-5.6-luna" → Model name specified in the code
# messages → Sends the conversation history
# Overall → Sends the conversation to the AI model

    ChatGPT_reply = response["choices"][0]["message"]["content"]
# ChatGPT_reply → Variable storing the AI response
# response → API response
# ["choices"] → Accesses the generated choices
# [0] → Selects the first choice
# ["message"] → Accesses the generated message
# ["content"] → Extracts the response text
# Overall → Extracts the AI-generated answer

    messages.append({"role": "assistant", "content": ChatGPT_reply})
# messages → Conversation-history list
# append → Adds an item to the list
# role → Defines the message role
# "assistant" → Identifies the AI response
# content → Stores the AI-generated response
# ChatGPT_reply → Generated response text
# Overall → Saves the AI response in the conversation history

    return ChatGPT_reply
# return → Sends a value back from the function
# ChatGPT_reply → AI-generated response
# Overall → Returns the response to the interface

demo = gradio.Interface(
    fn=CustomChatGPT,
    inputs="text",
    outputs="text",
    title="My Saving Assistant"
)
# demo → Variable storing the Gradio interface
# gradio → Gradio library
# Interface → Creates a web interface
# fn → Specifies the function to execute
# CustomChatGPT → Function connected to the interface
# inputs → Defines the input type
# "text" → Text input
# outputs → Defines the output type
# "text" → Text output
# title → Sets the interface title
# "My Saving Assistant" → Title displayed in the interface
# Overall → Creates the chatbot interface

demo.launch(share=True)
# demo → Gradio interface
# launch → Starts the interface
# share → Enables creation of a shareable link
# True → Enables the share option
# Overall → Launches the chatbot

# EXPLANATION:
# This code creates a custom chatbot using the OpenAI API.
# The API key is loaded from the .env file.
# A system message defines the chatbot's role.
# User and assistant messages are stored in the messages list.
# The OpenAI API generates the response.
# Gradio provides the chatbot web interface.
# share=True creates a shareable Gradio link.

# This cell creates and launches a custom AI chatbot
# using OpenAI API and Gradio.
```
